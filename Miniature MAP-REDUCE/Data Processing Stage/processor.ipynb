```python
import pandas as pd
import numpy as np
```

```python
# reading csv
df = pd.read_csv("../Data/Raw Data/Product_Reviews.csv",index_col=0)
df.head(10)
```

```python
df.isnull().sum()
```

```python
df = df.drop(columns=['Text','Summary','ProfileName'])
```

```python
df.head()
```

```python
df['Time'] = pd.to_datetime(df['Time'], unit='s')
```

```python
df.head()
```

```python
df.reset_index(drop=True, inplace=True)
```

```python
df.describe()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
# increase the data size

newdf = pd.concat(
    [
        df.assign(UserId=df["UserId"] + "_copy0"),
        df.assign(UserId=df["UserId"] + "_copy1"),
        df.assign(UserId=df["UserId"] + "_copy2")
    ],
    ignore_index=True
)
```

```python
newdf.head(100)
```

```python
newdf.shape
```

```python
newdf.to_csv(
    "../Data/Processed Data/processed_data.csv",
    index=False,
    encoding="utf-8"
)
```
